```python
import pandas as pd
import numpy as np
```

```python
rating=pd.read_csv(r"D:\MINI PROJECT\DATASET\MovieLensDataset20M\rating.csv")
movie=pd.read_csv("D:\MINI PROJECT\DATASET\MovieLensDataset20M\movie.csv")
tag=pd.read_csv(r"D:\MINI PROJECT\DATASET\MovieLensDataset20M\tag.csv")
genome_scores=pd.read_csv("D:\MINI PROJECT\DATASET\MovieLensDataset20M\genome_Scores.csv")
genome_tags=pd.read_csv("D:\MINI PROJECT\DATASET\MovieLensDataset20M\genome_tags.csv")
```

```python
datasets={'rating':rating,
          'movie':movie,
          'tag':tag,
          'genome scores':genome_scores,
          'genome_tags':genome_tags}

for name,i in datasets.items():
    print('Dataset:',name)
    rows, columns = i.shape
    print(f"Number of records: {rows}")
    print(f"Number of columns: {columns}")
    print('Any duplicate values:\n',i.isnull().sum())
    print('\n')
```

```python
print(movie['title'].head)
```

```python
extracted_content_df = movie['title'].str.extract(r'.*\(([^()]*)\)$')
extracted_content_series = extracted_content_df.iloc[:, 0]
numeric_years = pd.to_numeric(extracted_content_series, errors='coerce')
movie['Year of release'] = numeric_years.astype('Int64') 
print(movie)
```

```python
is_numeric_mask = pd.to_numeric(movie['Year of release'], errors='coerce').isna()
numeric_count = is_numeric_mask.sum()
print('Number of records with no year:',numeric_count)
```

```python
print("Records with no year")
no_year_records=movie[movie["Year of release"].isna()]
print(no_year_records)
```

```python
#cleaning title to extract title of movie and release year seperately
has_year=movie['Year of release'].notna()

movie['Clean Movie title']=np.where(has_year,
                                    movie['title'].str.split('(').str[0].str.strip(),
                                    movie['title'])
```
