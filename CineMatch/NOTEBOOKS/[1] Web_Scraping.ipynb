FINAL WEB SCRAPING FOR MISSED OUT MOVIES


```python
"""
Extend TMDb mapping for remaining MovieLens movies.

- Reads Movie_Cleaned_title.csv (all movies)
- Reads movie_tmdb_mapping.csv (existing ~15k mappings)
- Finds movieIds not yet processed
- For those, calls TMDb /search/movie and chooses the best match
- Appends new rows to movie_tmdb_mapping.csv
- No duplicates: each movieId appears at most once.
"""

import os
import time
import json
import math
from pathlib import Path

import requests
import pandas as pd
from urllib.parse import quote
from rapidfuzz import fuzz

# =========================
# CONFIG
# =========================

MOVIES_CSV = "D:\MINI PROJECT\DATASET\Movie_Cleaned_title.csv"      # full movie list
MAPPING_CSV = "movie_tmdb_mapping.csv"      # existing mapping
CHECKPOINT_EVERY = 500                      # save every N new rows
SLEEP_SECONDS = 0.7                         # delay between requests
TIMEOUT = 12
RETRIES = 3
BATCH_LIMIT = 2000      # for safety; set None to process all remaining

# Use your TMDb API key directly (simplest for now)
API_KEY = "656efa157ee0f47cae123095d9c0564e"

HEADERS = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64)"
}

# Force IPv4 for some networks (optional but often helpful)
try:
    import requests.packages.urllib3.util.connection as urllib3_cn
    urllib3_cn.HAS_IPV6 = False
except Exception:
    pass


# =========================
# Helpers
# =========================

def is_valid_year(y):
    if y is None:
        return False
    if isinstance(y, float) and math.isnan(y):
        return False
    try:
        int(y)
        return True
    except Exception:
        return False


def normalize_title(s: str):
    if s is None:
        return ""
    out = s.lower().strip()
    out = out.replace("&", " and ")
    out = " ".join(out.split())
    for ch in '.,:;!?"\'()[]{}':
        out = out.replace(ch, "")
    return out


def safe_search_title(title, year=None):
    """
    TMDb /search/movie with retries and simple backoff.
    Returns JSON or None.
    """
    q = quote(title)
    url = f"https://api.themoviedb.org/3/search/movie?api_key={API_KEY}&query={q}"
    if is_valid_year(year):
        url += f"&year={int(year)}"

    backoff = 1.0
    for attempt in range(1, RETRIES + 1):
        try:
            r = requests.get(url, headers=HEADERS, timeout=TIMEOUT)
            r.raise_for_status()
            return r.json()
        except Exception as e:
            print(f"[SEARCH error attempt {attempt}] title='{title}' year={year} -> {repr(e)}")
            if attempt < RETRIES:
                time.sleep(backoff)
                backoff *= 2
            else:
                return None
    return None


def top_k_candidates(js, k=5):
    if js is None:
        return []
    results = js.get("results") or []
    out = []
    for r in results[:k]:
        out.append({
            "tmdb_id": r.get("id"),
            "title": r.get("title"),
            "release_date": r.get("release_date"),
            "vote_count": r.get("vote_count") or 0,
            "popularity": r.get("popularity") or 0,
        })
    return out


def choose_match(ml_title, ml_year, candidates):
    """
    Same logic as before, but summarized:

    - exact_year: normalized titles equal AND years equal
    - near_exact: fuzzy similarity >=95 AND years equal
    - title_match: if ml_year missing, fuzzy similarity >=97 AND vote_count >= 500
    - else: ambiguous / no_match
    """
    if not candidates:
        return None, "no_match", None

    norm_ml = normalize_title(ml_title)

    top = candidates[0]
    tmdb_title = top["title"] or ""
    norm_top = normalize_title(tmdb_title)
    release_date = top["release_date"]
    tmdb_year = release_date[:4] if release_date else None

    # exact normalized title + same year
    if norm_ml == norm_top and is_valid_year(ml_year) and tmdb_year:
        try:
            if int(tmdb_year) == int(ml_year):
                return top["tmdb_id"], "exact_year", top
        except Exception:
            pass

    # fuzzy match + same year
    if is_valid_year(ml_year) and tmdb_year:
        score = fuzz.token_set_ratio(norm_ml, norm_top)
        if score >= 95 and int(tmdb_year) == int(ml_year):
            return top["tmdb_id"], "near_exact", top

    # no year on ML side: rely on strong title + popularity
    if not is_valid_year(ml_year):
        score = fuzz.token_set_ratio(norm_ml, norm_top)
        if score >= 97 and top["vote_count"] >= 500:
            return top["tmdb_id"], "title_match", top

    # otherwise ambiguous
    return None, "ambiguous", None


# =========================
# Main
# =========================

def main():
    # 1) Load full movie list
    if not os.path.exists(MOVIES_CSV):
        print(f"ERROR: {MOVIES_CSV} not found.")
        return

    movies = pd.read_csv(MOVIES_CSV)
    print("Total movies in Movie_Cleaned_title:", len(movies))

    # 2) Load existing mapping (or start new)
    if os.path.exists(MAPPING_CSV):
        mapping = pd.read_csv(MAPPING_CSV)
        mapping["movieId"] = mapping["movieId"].astype(str)
        processed_ids = set(mapping["movieId"].tolist())
        print("Existing mappings:", len(mapping))
    else:
        mapping = pd.DataFrame(columns=[
            "movieId", "ml_title", "ml_year",
            "tmdb_id", "tmdb_title", "tmdb_year",
            "vote_count", "match_confidence", "note"
        ])
        processed_ids = set()
        print("No existing mapping found. Starting fresh.")

    # 3) Compute remaining movies to process
    movies["movieId"] = movies["movieId"].astype(str)
    remaining = movies[~movies["movieId"].isin(processed_ids)].copy()
    print("Remaining movies to process:", len(remaining))

    if BATCH_LIMIT is not None:
        remaining = remaining.iloc[:BATCH_LIMIT].copy()
        print("Restricting this run to first", len(remaining), "remaining movies.")

    if remaining.empty:
        print("Nothing left to process. Exiting.")
        return

    new_rows = []
    processed_this_run = 0

    for idx, row in remaining.iterrows():
        movieId = row["movieId"]
        title = str(row.get("Clean Movie title") or row.get("title") or "").strip()
        ml_year = row.get("Year of release")

        print(f"\n[{movieId}] Searching TMDb for '{title}' year={ml_year} ...")

        js = safe_search_title(title, ml_year)

        if js is None:
            # network or TMDb issue
            print(f"[{movieId}] search_failed (TMDb unreachable)")
            new_rows.append({
                "movieId": movieId,
                "ml_title": title,
                "ml_year": ml_year,
                "tmdb_id": None,
                "tmdb_title": None,
                "tmdb_year": None,
                "vote_count": None,
                "match_confidence": "search_failed",
                "note": ""
            })
        else:
            candidates = top_k_candidates(js, k=5)
            chosen_id, conf, top = choose_match(title, ml_year, candidates)

            if top is not None:
                tmdb_title = top["title"]
                rdate = top["release_date"]
                tmdb_year = rdate[:4] if rdate else None
                vote_count = top["vote_count"]
            else:
                tmdb_title = None
                tmdb_year = None
                vote_count = None

            print(f"[{movieId}] match_confidence={conf}, tmdb_id={chosen_id}, tmdb_title={tmdb_title}")

            new_rows.append({
                "movieId": movieId,
                "ml_title": title,
                "ml_year": ml_year,
                "tmdb_id": chosen_id,
                "tmdb_title": tmdb_title,
                "tmdb_year": tmdb_year,
                "vote_count": vote_count,
                "match_confidence": conf,
                "note": ""
            })

        processed_this_run += 1

        # checkpoint
        if processed_this_run % CHECKPOINT_EVERY == 0:
            print(f"\nCheckpoint: processed this run = {processed_this_run}")
            new_df = pd.DataFrame(new_rows)
            combined = pd.concat([mapping, new_df], ignore_index=True)
            # just in case, drop duplicates on movieId, keep first
            combined = combined.drop_duplicates(subset=["movieId"], keep="first")
            combined.to_csv(MAPPING_CSV, index=False)
            print(f"Saved checkpoint mapping to {MAPPING_CSV}")

        time.sleep(SLEEP_SECONDS)

    # Final save
    new_df = pd.DataFrame(new_rows)
    combined = pd.concat([mapping, new_df], ignore_index=True)
    combined = combined.drop_duplicates(subset=["movieId"], keep="first")
    combined.to_csv(MAPPING_CSV, index=False)
    print("\nDone.")
    print("Total mappings now in file:", len(combined))


if __name__ == "__main__":
    main()
```

```python
import requests
import pandas as pd
import time
from urllib.parse import quote

API_KEY = "656efa157ee0f47cae123095d9c0564e"

BASE_URL = "https://api.themoviedb.org/3"

endpoints = [
    "movie/popular",
    "movie/top_rated",
    "movie/now_playing",
    "movie/upcoming",
    "trending/movie/week",
    "discover/movie"
]

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "application/json"
}

movies = []
seen_ids = set()

for endpoint in endpoints:

    print(f"\nFetching from endpoint: {endpoint}")

    for page in range(1, 13):

        url = f"{BASE_URL}/{endpoint}?api_key={API_KEY}&page={page}"

        try:
            response = requests.get(url, headers=headers, timeout=20)

            if response.status_code != 200:
                print("HTTP error:", response.status_code)
                continue

            data = response.json()

            if "results" not in data:
                print("API returned unexpected response:", data)
                continue

        except Exception as e:
            print("Connection failed:", e)
            continue

        for m in data["results"]:

            title = m.get("title")

            if not title:
                continue

            encoded_title = quote(title)

            search_url = f"{BASE_URL}/search/movie?api_key={API_KEY}&query={encoded_title}"

            try:

                # Retry mechanism
                for attempt in range(3):
                    try:
                        search_response = requests.get(search_url, headers=headers, timeout=20)
                        search_data = search_response.json()
                        break
                    except:
                        time.sleep(1)

                if "results" not in search_data or len(search_data["results"]) == 0:
                    continue

                best = search_data["results"][0]

                movie_id = best["id"]

                if movie_id in seen_ids:
                    continue

                seen_ids.add(movie_id)

                poster = best.get("poster_path")

                poster_url = None
                if poster:
                    poster_url = "https://image.tmdb.org/t/p/w500" + poster

                movies.append({
                    "movieId": movie_id,
                    "title": best.get("title"),
                    "overview": best.get("overview"),
                    "release_date": best.get("release_date"),
                    "rating": best.get("vote_average"),
                    "popularity": best.get("popularity"),
                    "vote_count": best.get("vote_count"),
                    "poster_path": poster_url
                })

                print("Saved:", best.get("title"))

                time.sleep(0.35)

            except Exception as e:
                print("Search failed:", e)


df = pd.DataFrame(movies)

df.drop_duplicates(subset=["movieId"], inplace=True)

df.to_csv("tmdb_new_movies.csv", index=False)

print("\nTotal movies scraped:", len(df))
```

```python
import pandas as pd
import requests
import time
from urllib.parse import quote

API_KEY = "656efa157ee0f47cae123095d9c0564e"

movies = pd.read_csv(r"D:\MINI PROJECT\DATASET\movies_extended.csv")

missing = movies[movies["genres"].isnull()]

print("Movies needing genre:", len(missing))

for idx, row in missing.iterrows():

    title = row["title"]

    if pd.isna(title):
        continue

    query = quote(title)

    url = f"https://api.themoviedb.org/3/search/movie?api_key={API_KEY}&query={query}"

    try:

        r = requests.get(url, timeout=10).json()

        if len(r["results"]) == 0:
            continue

        movie = r["results"][0]

        genres = movie.get("genre_ids")

        if genres:
            movies.loc[idx, "genres"] = str(genres)

        print("Updated:", title)

        time.sleep(0.25)

    except:
        print("Failed:", title)

movies.to_csv("D:/MINI PROJECT/DATASET/movies_final_merged.csv", index=False)

print("Genre fixing completed")
```

```python
#Convert genre ids to names 

import pandas as pd
import ast

movies = pd.read_csv("D:/MINI PROJECT/DATASET/movies_final_merged.csv")

genre_map = {
    28: "Action",
    12: "Adventure",
    16: "Animation",
    35: "Comedy",
    80: "Crime",
    99: "Documentary",
    18: "Drama",
    10751: "Family",
    14: "Fantasy",
    36: "History",
    27: "Horror",
    10402: "Music",
    9648: "Mystery",
    10749: "Romance",
    878: "Science Fiction",
    10770: "TV Movie",
    53: "Thriller",
    10752: "War",
    37: "Western"
}

def convert_genre_ids(x):

    if pd.isna(x):
        return x

    try:
        ids = ast.literal_eval(x)  # convert string "[28,12]" → list
        names = [genre_map.get(i, "") for i in ids]
        return "|".join(names)

    except:
        return x


movies["genres"] = movies["genres"].apply(convert_genre_ids)

movies.to_csv(
    "D:/MINI PROJECT/DATASET/movies_final_merged.csv",
    index=False
)

print("Genre conversion complete.")
```
